# Types of Arguments

1. Positional argument
2. Keyword argument
3. Default argument
4. Variable length argument

```python
# 1 . Posotional Arguments
def sub(a,b):
    return a+b
print(sub(10,20))
print(sub(20,1))
```

```python
# 2. Keyword Arguments
def wish(name,msg):
    print("Hello",name,msg)
wish(name="Meet",msg="Good morning")
wish("Meet",msg="good morning")
wish("Meet","Good morning")
wish(msg="Good morning",name="Meet")
```

```python
# 3. Default Arguments
def wish(name="Meet"):
    print("Hello",name,"How are you")
wish()
wish("Bhumit")
```

```python
# 4. Variable length Arguments
def sum(*s):
    total=0
    for i in s:
        total=total+i
    print(total)
sum(10)
sum(10,20,30,40)
sum(10,20,30,40,50,60)
print()
def sum(n1,*s):
    total=0
    for i in s:
        total=total+i
    print(total)
print()
sum(10)
sum(10,20,30,40)
sum(10,20,30,40,50,60)
```

```python
def sum(*s,n1):
    total=0
    for i in s:
        total=total+i
    print(total)

sum(10,20,30,40,n1=56)
```

```python
x=3 # global
def fun(a):
    x=5 # local
    return x*a
print(fun(5))
print(x) # global
```

```python
x=3 # global
def fun(a):
    global x # local to new global
    x=5
    return x*a
print(fun(5))
print(x)
```

# Nested Function

```python
def fun1():
    print("This is outer function")
    def fun2():
        print("This is inner function")
    fun2()
fun1()
```

```python
def fun1():
    print("This is outer function")
    def fun2():
        print("This is inner function")
    return fun2()
fun1()
```

```python
def fun1():
    print("This is outer function")
    def fun2():
        print("This is inner function")
fun2()
```

# For else, While else

```python
for i in range(6):
    print(i)
else:
    print("Moj Ma")
```

```python
x=1
while x<=6:
    print(x)
    x+=1
else:
    print("Moj Moj Moj")
```

```python
for i in range(6):
    print(i)
    if i==3:
        break
else:
    print("Moj Ma")
```
