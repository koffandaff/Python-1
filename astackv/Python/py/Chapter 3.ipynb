```python
#Different categories for user defined functions

#1. w/o argument and parameter, function without parameter and no return type

def wish():
    name=input("Enter a name: ")
    print(name)
wish()
```

```python
#2. w parameter and no return type

name=input("Enter a name: ")
def wish(name):
    print(name)
wish(name)

#2. w parameter and no return type w fixed name

def wish(name):
    print(name)
wish("Dhruvil")
```

```python
#3. w/o parameter and w return type

def wish():
    name=input("Enter a name: ")
    return name
wish()
```

```python
#4. w parameter and w return type

name=input("Enter a name: ")
def wish(name):
    return name
wish(name)
```

```python
#4. VARIABLE LENGTH 

def sum(*s):
    total=0
    for i in s:
        total=total+i
    print(total)
sum(10)
sum(10,20,30,40)
sum(31,134,314135,4356)
```

```python
#4. VARIABLE LENGTH 

def sum(n1,*s):
    total=0
    for i in s:
        total=total+i
    print(total)
sum(10)
sum(10,20,30,40)
sum(31,134,314135,4356)
```

```python
#4. VARIABLE LENGTH 

def sum(*s,n1):
    total=0
    for i in s:
        total=total+i
    print(total)
#sum(10)
#sum(10,20,30,40)
#sum(31,134,314135,4356)
sum(10,20,n1=24)
```

```python
#LOCAL AND GLOBAL VARIABLES

x=3
def fun(a):
    x=5
    return x*a
print(fun(5))
print(x)
```

```python
#LOCAL AND GLOBAL VARIABLES

x=3
def fun(a):
    global x
    x=5
    return x*a
print(fun(5))
print(x)
```

```python
# nested function
def fun1():
    print("This is outer function")
    def fun2():
        print("This is inner function")
    
fun1()
print(fun1())
```
