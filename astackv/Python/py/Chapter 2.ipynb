```python
# REVERSE NUMBER

n=int(input("Enter the number: "))
num = n
rev_num=0
while n!=0:
    ld=n%10
    rev_num=rev_num*10+ld
    n=n//10
print(f"Reverse number of {num} is {rev_num}")
```

```python
# PALINDROME NUMBER

n=int(input("Enter the number: "))
num = n
rev_num=0
while n!=0:
    ld=n%10
    rev_num=rev_num*10+ld
    n=n//10
print(f"Reverse number of {num} is {rev_num}")

if num==rev_num:
    print("This is a Palindrome Number")
else:
    print("Not a palindrome number tho")
```

```python
# ARMSTRONG NUMBER

n=int(input("Enter A number to check if its Armstrong: "))
num = n
sum=0
l=len(str(n))
while n!=0:
    ld=n%10
    sum=sum+(ld**l)
    n=n//10
    
if sum==num:
    print("Number is armstrong")
else:
    print("Not armstrong")
```

```python
# FIBONACCI SERIES

n=int(input("Enter a number of series characters: "))
n1=0
n2=1

if n==0 or n<0:
    print("Enter valid value")
elif n==1:
    print("Fibonacci series is:\n",n1)
else:
    print("Fibonacci Series:")
    print(n1)
    print(n2)
    for i in range(0,n-2):
        nth=n1+n2
        print(nth)
        n1=n2
        n2=nth
```
