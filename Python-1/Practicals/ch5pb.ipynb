if removal of that element makes the list balanced.

The list will be balanced if sum of even index elements is equal to the sum of odd index 
elements.

Example Input

Input 1:

A = [2,1,6,4]

Input 2:

A=[5,5,2,5,8]

Example Output

Output 1:

1

Output 2:

2

Explanation 1 :

After deleting 1 from list : [2,6,4]

(2+4) = (6)

Hence 1 is the only special element, so count is 1.

Explanation 2 :

If we delete A[0] or A[1], list will be balanced

(5+5)=(2+8)

So A[0] and A[1] are special elements, so count is 2

The Syracuse (also called Collatz or Hailstone) sequence is generated by starting with a natural 

number and repeatedly 

applying the following function until reaching 1:

 syr(x) = 1/2 if x is even; and 

syr(x) = 3x + 1 if x is odd 

For example, the Syracuse sequence starting with 5 is: 5, 16, 8, 4, 2, 1. 

Use of inbuilt function or math library is not allowed.


This sequence will always go to 1 for every possible starting value.

Write a program that 

 1.Gets a starƟng value from the user 

 2.Prints the Syracuse sequence for that starƟng valu

```python
def syracuse(num):
    while num > 1:
        if num %2 == 0:
            num = num // 2
        else:
            num = num*3 + 1
        print(num)

num = int(input("Enter a Number only: "))
print(f"For the number {num} its: ")
syracuse(num)
```

434. Sum of positive and negative numbers using Lambda

```python
l = [1,2,-2,-3,-4,2,1,2]

pos = sum(filter(lambda x: x>0, l))
print(pos)

neg = sum(filter(lambda x: x < 0, l))
print(neg)

print(f"The sum is: {pos} and negative: {neg}")
```

```python
L = [2, 4, 6, 8, 10]
n = len(L)
result = [] # Temporary list (used to maintain simple logic)
left, right = 0, n - 1

while left <= right:
    # Take the leftmost element
    result.append(L[left])
    left += 1

    # Take the rightmost element (only if pointers haven't crossed)
    if left <= right:
        result.append(L[right])
        right -= 1

# Copy result back to L (simulating in-place update)
L[:] = result 
print(L)
```
