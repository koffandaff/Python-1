```python
You own a pizzeria named Olly’s Pizzas and want to create a Python program to handle the customers and revenue. 
Create the following classes with the following methods:
Class Pizza containing
 1.init method: to ini alize the size (small, medium, large), toppings (corn, tomato, onion, capsicum, mushroom, olives, 
broccoli), cheese (mozzarella, feta, cheddar). Note: One pizza can have only one size but many toppings and cheese. (1.5 
marks)
Throw custom exceptions if the selects toppings or cheese not available in lists given above. (1 mark)
 2.price method: to calculate the prize of the pizza in the following way:
small = 50, medium = 100, large = 200
Each topping costs 20 rupees extra, except broccoli, olives and mushroom, which are exotic and so cost 50 rupees each.
Each type of cheese costs an extra 50 rupees. (1.5 marks)
 Class Order containing
 1.init method: to ini alize the name, customerid of the customer who placed the order (0.5 marks)
 2.order method: to allow the customer to select pizzas with choice of toppings and cheese (1 mark)
 3.bill method: to generate details about each pizza ordered by the customer and the total cost of the order. (2 marks)
*Note: A customer can get multiple pizzas in one order.
1.5 marks for creating appropriate objects of these classes and writing correct output.
```

````python
class InvalidSelection(Exception):
    def __init__(self, message):
        self.message = message

class Pizza:
    def __init__(self, size, toppings, cheeses):
        valid_toppings = ['corn', 'tomato', 'onion', 'capsicum', 'mushroom', 'olives', 'broccoli']
        valid_cheeses = ['mozzarella', 'feta', 'cheddar']
        
        self.size = size.lower()
        self.toppings = [t.lower() for t in toppings]
        self.cheeses = [c.lower() for c in cheeses]
        
        for t in self.toppings:
            if t not in valid_toppings:
                raise InvalidSelection(f"Topping '{t}' is not available.")
        
        for c in self.cheeses:
            if c not in valid_cheeses:
                raise InvalidSelection(f"Cheese '{c}' is not available.")

    def price(self):
       
        if self.size == 'small':
            cost = 50
        elif self.size == 'medium':
            cost = 100
        else:
            cost = 200
            
        
        for t in self.toppings:
            if t in ['broccoli', 'olives', 'mushroom']:
                cost += 50
            else:
                cost += 20
                
        
        cost += len(self.cheeses) * 50
        return cost

class Order:
    def __init__(self, name, customerid):
        self.name = name
        self.customerid = customerid
        self.pizzas = []

    def order(self):
        print(f"\n--- Adding a Pizza for {self.name} ---")
        sz = input("Enter size (small, medium, large): ")
        tops = input("Enter toppings separated by comma: ").split(',')
        chse = input("Enter cheeses separated by comma: ").split(',')
        
        try:
            p = Pizza(sz, tops, chse)
            self.pizzas.append(p)
            print("Pizza added to order!")
        except InvalidSelection as e:
            print(f"Error: {e.message}")
        
        
        cont = input("Add another pizza to this order? (yes/no): ")
        if cont.lower() == 'yes':
            self.order()

    def bill(self):
        print("\n" + "="*30)
        print(f"BILL FOR: {self.name} (ID: {self.customerid})")
        total = 0
        for i, p in enumerate(self.pizzas, 1):
            p_price = p.price()
            total += p_price
            print(f"Pizza {i}: {p.size} | Toppings: {', '.join(p.toppings)} | Cheese: {', '.join(p.cheeses)}")
            print(f"Cost: {p_price} rupees")
        print("-" * 30)
        print(f"TOTAL ORDER COST: {total} rupees")
        print("="*30)




cust_name = input("Enter Customer Name: ")
cust_id = input("Enter Customer ID: ")

customer_order = Order(cust_name, cust_id)
customer_order.order()
customer_order.bill()
````

```python
 words_with_length(length) — returns a list of all the words of length length
 starts_with(char1) — returns a list of all the words that start with char1
 ends_with(char2) — returns a list of all the words that end with char2
 palindromes() — returns a list of all the palindromes in the list
 only(str1) — returns a list of the words that contain only those le ers in str1

Make Required object for WordPlay class and test all the methods.

'apricot']

starts_with ('a') should return ['apple', 'apricot']

palindromes () should return ['malayalam', 'nayan']
only ('bna') should return ['banana']
avoids ('amkd') should return ['set', 'tuple', 'list']
```

````python
class WordPlay:
    def __init__(self, word_list):
        self.words = word_list

    def words_with_length(self, length):
        result = []
        for w in self.words:
            if len(w) == length:
                result.append(w)
        return result

    def starts_with(self, char1):
        result = []
        for w in self.words:
            if w.startswith(char1):
                result.append(w)
        return result

    def ends_with(self, char2):
        result = []
        for w in self.words:
            if w.endswith(char2):
                result.append(w)
        return result

    def palindromes(self):
        result = []
        for w in self.words:
            if w == w[::-1]:
                result.append(w)
        return result

    def only(self, str1):
        result = []
        for w in self.words:
            is_only = True
            for letter in w:
                if letter not in str1:
                    is_only = False
                    break
            if is_only:
                result.append(w)
        return result

    def avoids(self, str2):
        result = []
        for w in self.words:
            is_avoided = True
            for letter in w:
                if letter in str2:
                    is_avoided = False
                    break
            if is_avoided:
                result.append(w)
        return result

    def run_menu(self):
        print("\n--- WordPlay Menu ---")
        print("1. Length | 2. StartsWith | 3. EndsWith | 4. Palindromes | 5. Only | 6. Avoids | 7. Exit")
        choice = input("Enter choice: ")

        if choice == '1':
            L = int(input("Enter length: "))
            print(self.words_with_length(L))
        elif choice == '2':
            c = input("Enter character: ")
            print(self.starts_with(c))
        elif choice == '3':
            c = input("Enter character: ")
            print(self.ends_with(c))
        elif choice == '4':
            print(self.palindromes())
        elif choice == '5':
            s = input("Enter letters: ")
            print(self.only(s))
        elif choice == '6':
            s = input("Enter letters to avoid: ")
            print(self.avoids(s))
        elif choice == '7':
            return

       
        self.run_menu()




user_input = input("Enter words separated by space: ").split()


wp = WordPlay(user_input)
wp.run_menu()
````

````python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance_from_origin(self):
       
        d = math.sqrt(self.x**2 + self.y**2)
        return round(d, 2)

    def translate(self, dx, dy):
        self.x = self.x + dx
        self.y = self.y + dy
        return (self.x, self.y)

    def reflect_x(self):
        
        self.y = -self.y
        return (self.x, self.y)

    def distance(self, other_point):
        
        d = math.sqrt((other_point.x - self.x)**2 + (other_point.y - self.y)**2)
        return round(d, 2)




p1 = Point(1, 2)
print(f"Point ({p1.x}, {p1.y})")


print(f"Distance from origin - {p1.distance_from_origin()}")


old_x, old_y = p1.x, p1.y
new_pos = p1.translate(1, 1)
print(f"Translate method - point ({old_x},{old_y}) translated by (1,1) increment will be at {new_pos} now")


old_pos = (p1.x, p1.y)
reflected_pos = p1.reflect_x()
print(f"Reflect_x Method - Point {old_pos} after given reflection will be at {reflected_pos}")

p2 = Point(3, 4)
dist = p1.distance(p2)
print(f"Distance Method - distance between point ({p1.x},{p1.y}) and ({p2.x},{p2.y}) is {dist}")
````

Stacks and Queues. Write a class SQ that defines a data structure that can behave as both a queue (FIFO) or a stack (LIFO), 
There are five methods that should be implemented:

1. make a constructor with a valid parameter  
2. shift () returns the first element and removes it from the list. Also, use the custom(raise) excepon in this method.
3. unshift () "pushes" a new element to the front or head of the list
4. push() adds a new element to the end of a list
5. pop() returns the last element and removes it from the list
6. remove() returns the maximum element of the list and removes it from the list.
7. Create the object and call all methods of the SQ class.

````python
# OUT of time so :)

class EmptyListError(Exception):
    def __init__(self, message):
        self.message = message

class SQ:
    def __init__(self, initial_list=None):
        if initial_list is None:
            self.data = []
        else:
            self.data = initial_list

    def shift(self):
        if len(self.data) == 0:
            raise EmptyListError("Cannot shift: The list is empty")
        return self.data.pop(0)

    def unshift(self, element):
        self.data.insert(0, element)

    def push(self, element):
        self.data.append(element)

    def pop(self):
        if len(self.data) == 0:
            raise EmptyListError("Cannot pop: The list is empty")
        return self.data.pop()

    def remove(self):
        if len(self.data) == 0:
            raise EmptyListError("Cannot remove max: The list is empty")
        max_val = max(self.data)
        self.data.remove(max_val)
        return max_val

# --- Test Case ---

# 1. Create Object
my_sq = SQ([10, 20, 30])
print(f"Original list: {my_sq.data}")

# 2. Push (Add to end)
my_sq.push(40)
print(f"After push(40): {my_sq.data}")

# 3. Unshift (Add to front)
my_sq.unshift(5)
print(f"After unshift(5): {my_sq.data}")

# 4. Pop (Remove from end - Stack behavior)
last = my_sq.pop()
print(f"Popped element: {last}")
print(f"List after pop: {my_sq.data}")

# 5. Shift (Remove from front - Queue behavior)
try:
    first = my_sq.shift()
    print(f"Shifted element: {first}")
    print(f"List after shift: {my_sq.data}")
except EmptyListError as e:
    print(e.message)

# 6. Remove (Remove maximum)
maximum = my_sq.remove()
print(f"Removed max element: {maximum}")
print(f"Final list: {my_sq.data}")
````
