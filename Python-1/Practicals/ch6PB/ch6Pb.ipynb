## 468
Write a function display_oddLines() to display odd number lines from the text file. Consider the following lines for the file – friends.txt.

Friends are crazy, Friends are naughty !

Friends are honest, Friends are  best !

Friends are like keygen, friends are like license key !

````python
def display_oddLines():
    with open('friends.txt','r') as f:
        lines = f.readlines()
        od_lines = [lines[line] for line in range(len(lines)) if line%2 != 0]
        
        for i in od_lines:
            print(i)
        
display_oddLines()
````

## 474

Write a python program to search for a string in text files.

````python
def search_files(files,text):
    for file in files:
        try:
            with open(file,'r') as f:
                lines = f.readlines()
                found_count = 0
                print(f"\n\n--------------Searching in file {file} for text: {text} -----------\n\n")
                for line in range(len(lines)):
                    if text in lines[line]:
                        print(f"Match Found For {text} in {file} at line {line}")
                        found_count += 1
                if found_count <= 0:
                    print(f"No match found for text: {text} in File: {file}" )
        except Exception as e:
            print(f"Error While operating on file: {file} , Erro: {e}")
            continue




Input_files = input("Enter the file Name if more than one file then seprate it with comma (,): ")
files = Input_files.split(',')
print(files)
search = input('Whatever string You want to search for in those files: ')

search_files(files,search)
````

## 475 

Write a “pager” program. Your solution should prompt for a filename, and display the text file 25 lines at a time, pausing 

each time to ask the user to enter the word “continue”, in order to show the next 25 lines or enter the word “stop” to 

close the file.

````python
file = input("Enter file name to read: ")

with open(file, 'r') as f:
    lines = f.readlines()

No = 1
cNo = 0
buff = 25
tlines = len(lines)

while True:

    if cNo >= tlines:
        print("========== End of File ==========")
        break

    for i in range(cNo, min(buff, tlines)):
        print(f"Line No {No}: {lines[i].strip()}")
        No += 1

    cNo = buff
    buff += 25

    prompt = input("Enter 'continue' to continue or 'stop' to terminate: ")
    if prompt.lower() == 'stop':
        print("Exiting the pager")
        break
````

## 478

write a python program to read a text file “Story.txt” and print only word starting with ‘I’ in reverse order.

Example: If value in text file is : ‘INDIA IS MY COUNTRY’

Output will be: ‘AIDNI SI MY COUNTRY’

````python
with open('story.txt', 'r') as f:
    lines = f.readlines()

with open('story_modified.txt', 'w') as f:
    for line in lines:
        words = line.split()
        new_words = []

        for word in words:
            if word.startswith('I'):
                new_words.append(word[::-1])  
            else:
                new_words.append(word)

        print(new_words)
        new_line = " ".join(new_words)
        print(new_line)
        f.write(new_line + "\n")


````

````python
with open('480.txt','r') as f:
    lines = f.readlines()
    
new_lines = []

for line in lines:
    if '#' in line:
        words = line.split()
        new_words = []
        for word in words:
            if word.startswith('#'):
                break
            else:
                new_words.append(word)
        if len(new_words) <= 0:
            continue
        new_line = " ".join(new_words)+'\n'
        new_lines.append(new_line)
                
        
    else:
        new_lines.append(line)
print(new_lines)

with open('480text2.txt','w') as f:
    f.writelines(new_lines)
````
