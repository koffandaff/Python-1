311. Calculate the number of uppercase, lowercase letters, and digits

```python
s = "Hello Pyth@n is 100% easy"
upper, lower, digits = 0, 0, 0

for char in s:
    if char.isupper():
        upper += 1
    elif char.islower():
        lower += 1
    elif char.isdigit():
        digits += 1

print(f"Uppercase letters: {upper}")
print(f"Lowercase letters: {lower}")
print(f"Digits: {digits}")
```
