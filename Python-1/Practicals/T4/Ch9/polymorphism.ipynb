### OPERATOR OVERLOADING FOR COMPLEX NO.

````python
class Complex:
    def __init__(self,r,i):
        self.real = r
        self.imaginary = i

    def __add__(self,other):
        real = self.real + other.real
        ima = self.imaginary + other.real

        return f'{real}+{ima}i'


c1 = Complex(1,4)
c2 = Complex(3,5)

#BEFORE __ADD__ METHOD print(c1+c2) TypeError: unsupported operand type(s) for +: 'Complex' and 'Complex'
print(c1+c2)
````

### FOR Comparator

````python
class A:
    def __init__(self,a):
        self.a =a 
    def __gt__(self,other):
        if self.a>other.a:
            return f'{self.a} is higher'
        else:
            return f'{other.a} is higher'
        


x= A(3)
y = A(4)
print(x>y)
````

## Method Overriding [Inheritance]

* **Types**
    1. Single Inhertiance:  single Parent -> single Child
    2. Multiple Inheritance: Two Parent -> single Child
    3. MultiLevel Inheritance: Grand Parent -> Parent -> Child
    4. Hierarchical Inheritance: NUH-UH
    5. Hybrid Inheritance: As the name suggests dumb-ahh

#### Single Inheritance

````python
# Basic overview
class A:
    @staticmethod
    def sum(a,b):
        print(a+b)

class B(A):
    @staticmethod
    def mul(a,b):
        print(a*b)

obj = B()

obj.mul(3,5)
obj.sum(3,5)
````

````python
# Method Overriding using Single Inheritance
class Parent:
    def sleep(self):
        print("Sleep at 9")
    
class Child(Parent):
    def sleep(self):
        print("Sleep at 3")
        super().sleep()
        
    def eat(self):
        print("Eat Protien")


obj =  Child()
obj1 = Parent()

# obj1.sleep()
obj.sleep()
obj.eat()
````

### Multiple Inheritance

````python
class Father:
    def Whoami(self):
        print("I am the Father")
    def Ha(self):
        print("HOHOHOHOHOHOH")

class Mother:
    def Whoami(self):
        print("I am the Mother")
    def Yo(self):
        print("YOGKNLKJBPLKPOIBH")

class Son(Mother,Father): # Order Matters
    def Whoami(self):
        # super().Whoami()
        super().Whoami()

        print("I am son")
    
print(Son.mro())


s = Son()
s.Whoami()
s.Ha()
s.Yo()

````

````python
class GFather:
    def Whoami(self):
        print("I am the Grand Father")
    def Ha(self):
        print("HOHOHOHOHOHOH")

class Father(GFather):
    def Whoami(self):
        print("I am the Father")
        super().Whoami()
    def Yo(self):
        print("YOGKNLKJBPLKPOIBH")
        

class Son(Father): # Order Matters
    def Whoami(self):
        print("I am son")
        super().Whoami()
    
print(Son.mro())


s = Son()
s.Whoami()
s.Ha()
s.Yo()
````

### Hirerachical 

````python
class Parent:
    def sleep(self):
        print('Sleep at 6') 
    def Whoami(self):
        print("I AM HIM")

class Child1(Parent):
    def sleep(self):
        print('Sleep at 12')
        super().sleep()

class Child2(Parent):
    def sleep(self):
        print('sleep at 10')
        super().sleep()

class Child3(Parent):
    def sleep(self):
        print('Sleep at 1')
        super().sleep()


c1 = Child1()
c2 = Child2()
c3 = Child3()

print(f'ORDER OF THIS INHERITANCE: CHILD1: {Child1.mro()}')
# OVerriding 
c1.sleep()
c2.sleep()
c3.sleep()

print(f'------- ACCESSING PARENT METHOD THROUGH EACH CHILD--------')
c1.Whoami()
c2.Whoami()
c3.Whoami()
        
        
````

### Hybrid

````python
# 

class Parent1:
    def Part(self):
        print('Im Head')

    def u1(self):
        print('P1')
    
class Parent2:
    def Part(self):
        print('Im Torso')

    def u2(self):
        print('P2')

class Child(Parent1,Parent2):
    def Part(self):
        print('I  The blessed one')
        super().Part()
    def u3(self):
        print('C')

class GC1(Child):
    def u4(self):
        print('GC1')

class GC2(Child):
    def u5(self):
        print('GC2')



g1 = GC1()

# for GC1
g1.Part()
g1.u1()
g1.u2()
g1.u3()
g1.u4()


g2 = GC2()

# For GC2

g2.Part()
g2.u1()
g2.u2()
g2.u3()
g2.u5()
````
