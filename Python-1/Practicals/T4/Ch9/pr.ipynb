```python
from abc import ABC, abstractmethod

class Employee(ABC):

    def get_employee(self): # id, name , salary
        self.id = input('Enter your id: ')
        self.name   = input("Enter Your name: ")
        self.salary = int(input('Enter Your salary: '))
    def display_detail(self):
        print(f'id: {self.id} | Name: {self.name} | salary: {self.salary}')
    def get_salary(self):  # store salary
        calc = self.salary
        return calc
    
    @abstractmethod
    def emp_id(self):
        pass


class Perk(Employee):
    def calc_perk(self):
        calc = self.salary
        self.hra = (calc*17)/100
        self.da = (calc*35)/100
        self.pf = (calc*12)/100
        self.totalperks()

    def print_details(self):
        self.display_detail()
        print('DA: ', self.da)
        print(f'hra: {self.hra}')
        print(f'pf: {self.pf}')
    def totalperks(self):
        self.totalperk =  self.da + self.hra - self.pf

    def emp_id(self):
        print('Emp Id')

class NetSalary(Perk):
    def get_total(self):
        self.calc_perk()
        self.tsalary = self.salary + self.totalperk
        
    def display(self): # id name salary da hra pf net salary
        self.print_details()
        print(f'Net salary: {self.tsalary}')
    def emp_id(self):
        print('Emp Id')

obj = NetSalary()
obj.get_employee()
obj.display_detail()
obj.get_total()
obj.display()
```
