```python
import matplotlib.pyplot as plt
import numpy as np

durations = [40, 50, 45, 55, 60, 30, 40, 50, 45, 55, 60, 30, 40, 50, 45, 55, 60, 30, 40, 50, 45, 55, 60, 30, 40, 50, 45, 55]
calories = [200, 250, 220, 270, 300, 150, 200, 250, 220, 270, 300, 150, 200, 250, 220, 270, 300, 150, 200, 250, 220, 270, 300, 150, 200, 250, 220, 270]

# 1. Weekly Averages
weekly_calories = []
for i in range(0, len(calories), 7):
    week = calories[i:i+7]
    weekly_calories.append(sum(week) / len(week))

# 2. Creating the 3x2 Grid
plt.figure(figsize=(12, 12))

# Subplot 1: Line Plot
plt.subplot(3, 2, 1)
plt.plot(durations, 'b-o', markersize=8)
plt.title("Daily Workout Durations")

# Subplot 3: Bar Chart
plt.subplot(3, 2, 3)
plt.bar(range(len(durations)), durations, color='blue')
plt.title("Daily Workout Durations (Bar Chart)")

# Subplot 4: Histogram
plt.subplot(3, 2, 4)
plt.hist(durations, bins=10, color='skyblue', edgecolor='black')
plt.title("Workout Durations Distribution")

# Subplot 5: Pie Chart
plt.subplot(3, 2, 5)
plt.pie(weekly_calories, labels=['W1', 'W2', 'W3', 'W4'], startangle=90, autopct='%1.1f%%')
plt.title("Weekly Calories Distribution")

plt.tight_layout()
plt.show()
```
