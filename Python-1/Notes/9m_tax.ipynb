# 9m que Annual Income and Tax Calculator

```python
# Ts so tuff

# constant values
PT = 200
TA = 900

#Defining Values
grade = input("Enter the Grade(A,B,C,D,E,F): ")
city = int(input("Enter the tier of your city [1,2,3]: "))


# Defining summation of predefined values for the Starter
monthly_income = TA - PT

if grade.lower() == 'a':
    basic_pay = 60000
    other_allowance = 8000
elif grade.lower() == 'b':
    basic_pay = 50000
    other_allowance = 7000
elif grade.lower() == 'c':
    basic_pay = 40000
    other_allowance = 6000
elif grade.lower() == 'd':
    basic_pay = 30000
    other_allowance = 5000
elif grade.lower() == 'e':
    basic_pay = 20000
    other_allowance = 4000
elif grade.lower()  == 'f':
    basic_pay = 10000
    other_allowance = 3000
else: 
    print("Enter a Valid Grade")

if city == 1:
    HRA = basic_pay *0.3
elif city == 2:
    HRA = basic_pay*0.2
elif city == 3:
    HRA = basic_pay*0.1
else:
    print("Enter a Valid City Tier 1 to 3")
    
montly_income = 0.5*basic_pay + basic_pay + other_allowance - 0.11*basic_pay + HRA
print(f"Monthly Income is: {montly_income}")

print(f"Annual Income: {montly_income*12}")
```
