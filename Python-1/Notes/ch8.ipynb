```python
# Exception Handling
def div (a,b):
    print(a / b)
try:
    div(5,0)
except:
    print("Bhul avti bandh karvo")
```

```python
# Else get print when there is no exception
def div (a,b):
    print(a / b)
try:
    div(5,1)
except:
    print("Bhul avti bandh karvo")
else:
    print("sachu tho hatu bhai")
```

```python
# finally Gets printed always
def div (a,b):
    print(a / b)
try:
    div(5,2)
except:
    print("Bhul avti bandh karvo")
else:
    print("sachu tho hatu bhai")
finally:
    print("always print")
```

```python
# to print the exception message
def div (a,b):
    print(a / b)
try:
    div(5,0)
except Exception as e:
    print(e)
else:
    print("sachu tho hatu bhai")
finally:
    print("always print")
```

```python
# Custom Exception
class Mahi(Exception):
    def __init__(self, msg="predefined message from mahi"):
        self.msg = msg
        print(self.msg)

def add(a,b):
    if type(a) == str:
        raise NameError("a str ma na chale nikal")
    if type(b) == str:
        raise Mahi()
    return a + b
add(10.5,20)
```

##### Types of constructure 
* Constructure with argument (Parametrized constructure)
* None parameterized constructure (no arg other then self)
* Default constructure


```python
# None parameterized
class check:
    fav="Apple"
    def __init__(self):
        self.fav="orange"
        self.show()
    def show(self):
        print(self.fav)
a=check()
```

```python
# default parametrized
class check:
    fav="Apple"
    def __init__(self):
        # self.fav="orange"
        self.show()
    def show(self):
        print(self.fav)
a=check() 
```

```python
# Destructor 
class Employee:
    increment=2
    def __init__(self,name,age,salary):
        self.name=name
        self.age=age
        self.salary=salary
    def detailprint(self):
        # self.salary=self.salary*self.increment
        print(f'{self.name} salary is {self.salary*self.increment} at the age of {self.age}')
A=Employee("Ram",25,40000)
A.detailprint()
b=Employee("Manish",24,50000)
b.detailprint()
del A           # To del an obj of a class
A.detailprint()
```

```python
# Class method and static method
class Student:
    def __init__(self,name,rollno):
        self.name=name
        self.rollno=rollno
    def show(self):
        print(self.name)
tej=Student("Tejas",10)
print(tej.name)
tej.show()
```

```python
class Student:
    def __init__(self,name,rollno):
        self.name=name
        self.rollno=rollno
    @staticmethod    
    def show():
        print("How are u?")
tej=Student("Tejas",10)
print(tej.name)
tej.show()
```

```python
class Myclass:
    a=5
    def __init__(self,x):
        self.x=x
        print(self.x)
    def func1(self):
        print(self.x)
    @classmethod
    def method1(cls):
        print(cls.a)
    @staticmethod
    def method2(y):
        print(y)
w=Myclass(9)
w.func1()
w.method1()
w.method2(100)
```
