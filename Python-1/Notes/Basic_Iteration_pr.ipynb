
# WAP to find sum of N numbers
---

```python
n = int(input("Enter the Number of summation: "))
sum = 0

for i in range(1,n+1): #using 1 as the start to improve time complexity
    sum += i
    
print(f"The summation of {n} is: {sum} \n The average is: {sum/n}")
```

# WAP to Read 3 numbers [a,b,c] and check how many numbers available between a and b which is divisible by c

```python
# BRute force it's not optimized
a = int(input("Enter the value ofA: "))
b = int(input("Enter the value of B: "))
c = int(input("Enter the value of c: "))

an = b-a # assigning the availabe numbers
count = 0 # setting the count to 0
for i in range(1,an+1):
    if i%c == 0: # checking if the number is divisible by c
        count +=1

print(f"The Total availabe numbers between {a} and {b} are: {an} \nThe amount of numbers divisible by c = {c} is: {count}")
```

# WAP to print multiplication table of given number by user 

```python
no = int(input("Enter the number for Multiplication table: "))

for i in range(1,11):
    
    if i == 1:
        print("Multplication Table: \n--------------------")
    print(f"| {no} x {i} = {no*i}  |")
    

# using while
j=0
print("\n\n\n")
while j <=10:
    print(f"| {no} x {j} = {no*j}  |")
    j+=1
```

# WAP to Find out Factorial of given number 

```python
f = int(input("Enter a number: "))
fact = 1
if f<0:
    print("Enter a valid number")
elif f ==0 or f==1:
    print("The factorial is 1")
else:
    for i in range(2,f+1):
        fact = fact*i
    print(fact)
    
```
