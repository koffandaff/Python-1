# Arguments

## Types of Argument:
1. Positional
2. Keyword
3. Default
4. variable len

---
### Positional Argument

```python
def pos(a,b):
    print(a,b)
pos(10,20) # Positional Argument means that 10 will go to a and 20 to b
pos(20,10) # Positional Argument means that 20 will go to a and 10 to b
```

---
### Keyword Argument 

```python
def wish(name,msg):
    print(name,msg)
wish(name="Vraj",msg="Vello") # will work fine 
wish("Vraj", "Niggesh") # Vraj to name Niggesh to msg
wish(msg="Vello", name="Vraj") # will work fine as i defined it 
wish("Vello", "Vraj") # here we are f**ed as vello will go to name and vraj to msg
wish("Parth", msg="YO") # it'll work just fine
# wish(name="Parth", "Yo")  will hit the error --> positional argument follows keyword argument
```

---
### Default Argument 

```python
def wish(name="Dhruvil"):
    print(name)
    
wish()
wish("Vraj")
```

---
### Variable Length Argument

```python
def sum(*s): # can take any amount of argument
    total = 0
    for i in s:
        total = total +i
    print(total)
    
sum(10)
sum(17,10)
sum(10,1231,22,22,33)
```

##### case 2

```python
def sum(n1, *s): # first argument as n1 and the others as s
    total = 0
    for i in s:
        total = total +i
    print(total)
    
sum(10)
sum(17,10)
sum(10,1231,22,22,33)
```
