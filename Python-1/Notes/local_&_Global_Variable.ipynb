# Local and Global Variable

```python
x = 3 # defining a global variable
def fun(a):
    x = 5 # Initially a local Variable as it is in function
    return x*a 
print(fun(5)) # will return 25
print(x) # will return 3 as it's our global variable
```

```python
x = 3 # defining a global variable
def fun(a):
    global x # Indicating X as our global var in a function
    x = 5 # Initially a local Variable as it is in function but we changed it into global in the above loc
    return x*a 
print(fun(5)) # will return 25
print(x) # will return 5 as it's our new global variable
```
