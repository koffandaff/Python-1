# WAP To check Wheter the Given Number is a Happy Number or not

```python
# Happy NUmber

num = int(input("Enter a number: "))
op_val = num # assigning it to a temp variable so the orignial can be used later 

# We gotta break the loop at 1 or 4 to avoid infinite loop 
# 1 is happy, 4 is not happy
while op_val != 1 and op_val != 4:
    sum = 0 # Intializing sum in the loop as we need to refresh it every time
    s_op_val = str(op_val) # COnverting Integer to the string so i can split number
    for i in s_op_val:
        sum = sum + int(i)**2 # COnverting i to int for this operation 
        op_val = sum # assiging the sum to our operating variable for the next iteration 


# Checking the value of a to Detemine if it was a happy number or not 
if op_val == 1:
    print(f"The number: {num} is a Happy Number!")
else: 
    print(f"THe number: {num} is not a Happy number")

```
