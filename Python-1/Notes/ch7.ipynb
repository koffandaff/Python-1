```python
%%writefile mymodule.py
def sum(a, b):
    print(a + b) 
def sub(a, b):
    print(a - b) 
def mul(a, b):
    print(a * b)
def div(a, b):
    print(a / b)
def pow(a, b):
    print(a ** b)
```

```python
import mymodule as m
m.sub(10,20)
m.div(10,2)
m.floor(10,2)
```

```python
from mymodule import *
sub(10,20)
```

```python
from mymodule import sum, sub
sum(10,20)
sub(10,20)
mul(10,20)  
```

```python
from mymodule import sum as su , sub as s
s(10,5)
su(10,10)
```

```python
%%writefile string_fun.py
def func(s):
    print("Lenght is:",len(s))
    print("Upper is:", s.upper())
    print("Lower is:", s.lower())
    print("Captialize is:", s.capitalize())
    print("Isupper:", s.isupper())
    print("Islower:", s.islower())
```
