```python
# WAP to copy odd lines of one file into another file name as odd.txt
f=open('test3.txt','r')
f1=open('odd.txt','w')
x=f.readlines()
for i in range(len(x)):
    if i%2==0:
        f1.write(x[i])
f.close()
f1.close()
```

```python
#WAP to compare 2 file and display the differences.
f=open('compare1.txt','r')
f1=open('compare2.txt','r')
x=f.read()
y=f1.read()
line,pos=1,1
for i in range(len(x)):
    if x[i]==y[i]:
        pos+=1
        if x[i]=="\n":
            line+=1
            pos=1
    else:
        print("line:",line,"position:",pos)
        pos+=1
f.close()
f1.close()
```

```python
#WAP to display all lines of a file except those that have "#" sign(means comments).

# f=open('test4.txt','r')
# f1=open('nts.txt','w')
# li=f.readlines()
# for i in li:
#     if i.startswith('#'):
#         continue
#     else:
#         if '#' in i:
#             idx=i.index('#')
#             f1.write(i[0:idx]+'\n')
#         else:
#             f1.write(i)
# f.close()
# f1.close()


f=open('test4.txt')
f1=open('nts.txt','w')
while True:
    line=f.readline()
    if len(line)!=0:
        if line[0]=='#':
            continue
        else:
            if '#' in line:
                for i in range(1,len(line)):
                    if line[i]=='#':
                        f1.write(line[0:i]+'\n')
                        break
            else:
                f1.write(line)
    else:
        break
f.close()
f1.close()
```

```python
# WAP to count words characters and space from given txt file.
f=open('test5.txt','r')
data=f.read()
chars=0
spaces=0
words=len(data.split())
for i in data:
    if i ==' 'and i !='\n':
        spaces+=1
    if i.isalpha():
        chars+=1
print("Words:",words)
print("Characters:",chars)
print("Spaces:",spaces)
f.close()

# Different Approach
print()
f=open('test5.txt','r')
b=f.read()
x=b.split()
print("Words:",len(x))
ch=len(b)-b.count(' ')-b.count('\n')
print("Characters:",ch)
space=b.count(' ')
print("Spaces:",space)
```
