# Iteration

---

## For

```python
for i in range(500000):
    if i>5:
        break
    print(f"{i}. Ballerina cappuchina")
```

```python
l = ["Rono", "Pessi", "Macdonaldo", "Gando Norris"]
name = "Max Verstappen"

print("For list:")
for i in l:
    print(i)

print("\n ----------------------")
print("for String:")
for i in name:
    print(i)

print("\n ----------------------")

d = {1: "Hey", "hi": "Hello", 3: "Holla", 4: "YOo"}
print("\n For Dictionary: ")
for i in d:
    
    print(f"Key: {i} | Value: {d[i]}")
```

---

## While 

```python
x = 0
while True:
    if x > 5:
        break
    print(x)
    x += 1
```

# Nested for

```python
for i in range(3):
    
    for j in range(4):
        print(f"i: {i}, j: {j}")
        
```
