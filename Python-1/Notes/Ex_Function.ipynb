### odd/even

```python
def Findz(val):
    if val%2 ==0:
        print("It's Even")
    else:
        print("Duh! It's odd")

print("Enter 3 integers")
a = int(input("Enter Int1: "))
b = int(input("Enter Int2: "))
c = int(input("Enter Int3: "))

print("REsults::")

Findz(a)
Findz(b)
Findz(c)
```

### Making of a calculator
___

```python
 

def calc(op,a,b):
    if op == '+':
        sum = a+b
    elif op == '-':
        sum = a - b
    elif op == "*":
        sum = a*b
    else:
        print("Enter a valid operation")
    return sum

a = int(input("Enter the first Integer: "))
b = int(input("Enter the second Integer: "))
op = input("Enter the Operation You want to perform [+,-,*]: ")

sum = calc(op,a,b)
print(f"The result is: {sum} \n And the Type of sum is: {type(sum)}" )
```
