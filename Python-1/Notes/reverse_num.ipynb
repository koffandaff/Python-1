# WAP to FIndout reverse number of the input

```python
ori =int(input("Enter your number: ")) 
n = ori # temporary variable which will be used to perform the operations 
rev_num = 0

while n!=0:
    ld = (n%10) # storing the last digit
    n = (n//10) # storing the number except last digit for next iteration 
    rev_num = rev_num*10 + ld # Storing the reverse number 
    

print("The Reverse number is: ",rev_num)
```
