### functionw with no Parameter and No Return Typw

```python
def Zap():
    print("Wyd, Huzz!")
def Old_hag():
    print("Get off old hag!")
def Unc():
    print("You have reached the unc status!")

gen = input("Enter the Generation You are in  [X,Z]")
if gen.lower() == 'z':
    Zap()
elif gen.lower() == 'x':
    Unc()
else:
    Old_hag()
```

---
### Function with Parameter and no return Type

```python
def Zap(name):
    print(f"Wyd, {name} The Huzz!")
def Old_hag(name):
    print(f"Get off old hag! mf {name}")
def Unc(name):
    print(f"WOMP WOMP {name} \nYou have reached the unc status! ")

name = input("Enter Your name: ")
gen = input("Enter the Generation You are in  [X,Z]")

if gen.lower() == 'z':
    Zap(name)
elif gen.lower() == 'x':
    Unc(name)
else:
    Old_hag(name)
```
