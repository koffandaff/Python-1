# Palindrome:


```python
ori =int(input("Enter your number: ")) # Assigning the orignal value for final evaluation
n = ori # assigning the temp vairable for operations 
m = 10 # will be used in fomula 
rev_num = 0 

while n!=0:
    ld = (n%10) # Finding the last digit of the current number will get 5 from 135
    n = (n//10) # upadting the new number eg 13
    rev_num = rev_num*m + ld # storing the last digit using this formula
    

print(rev_num)

if ori == rev_num:
    print("The number is palindrome")
else:
    print("The number is not Palindrome")
    
```

# Armstrong

```python
n = int(input("Enter a number to verify if it's armstrong: "))
ori =n # Assigning the orignal value for final evaluation

count = len(str(n)) # finding the length of the integer by convrerting it into string 

arm_n = 0 # assigning the armstrong varaiable 
while n!=0:
    ld = (n%10) # taking out the last digit from the whole integer
    
    arm_n = arm_n + ld**count # Formula of armstrong
    
    
    
    n = (n//10) 
    
if arm_n== ori:
    print("The number is armstrong")
else:
    print("Not Armstrong")

```
