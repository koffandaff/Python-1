# Patterns 

### Left Based Triangle 

```python
for i in range(6):
    print("*" *i)
print()
print("-----------------------------------")

# Second Method 

row = int(input("Enter The number of rows"))

for i in range(1,row+1):
    for j in range(1,i+1):
        print("*", end=" ")
    print()

print()
print("-----------------------------------")

# third method 
for i in range(row):
    print("*", end=" ")
    for j in range(i):
        print("*", end =" ")
    print()
```

# Left-Based Numbered Triangle


```python
for i in range(1,row+1):
    for j in range(1,i+1):
        print(i, end=" ")
    print()
```

## REversed Left-Based Pyramid

```python
row = int(input("Enter The number of rows"))
print(row)

for i in range(1,row+1):
    for j in range(row-i+1):
        print("*", end=" ")
    print()
```

# Right sided Pyramid

```python
row = int(input("Enter The number of rows"))
print(row)

for i in range(1,row+1):
    for j in range(row-i+1):
        print(" ", end=" ")
    for k in range(1, i+1):
        print("*", end=" ")
    print()
```
