### 2.while loop 

```python
x=0
while x<6:
    print (x)
    x=x+1
```

### 3.Nested loop 

```python
for i in range(3):
    for j in range(3):
        print("i=",i,"j=",j)
```

### break

```python
for i in range(10):
    if i==7:
        print("enough")
        break
    print(i)
```

### continue 

```python
for i in range(10):
    if i==7:
        print("iteration is enough")
        continue
    print(i)
```

### Write a python program to check even year is leap year or not. 

```python
y = int(input("Enter a year:"))

if (y%4==0 and y%100!=0) or (y%400==0):
    print("Year is Leap year")
else:
    print("Year is not Leap year")
```

### Write a python program to perform arithmetic operations according to choice given by user. 

```python
a=int(input("Enter value 1:"))
b=int(input("Enter value 2:"))
c=input("Enter the choice:")

if c=='+':
    print("Addition:",a+b)
elif c=='-':
    print("subtraction:",a-b)
elif c=='*':
    print("Multiplication:",a*b)
elif c=='%':
    print("Modulo:",a%b)
elif c=='/':
    print("division:",a/b)
else:
    print("Please enter valid choice")
```

### Write a python program to calculate the electricity bill (no of units decided by user) .

```python
unit = int(input("Enter Units:"))
amount=0

if unit<=100:
    print("amount:",amount)
elif unit>=100 and unit<=200:
    amount=(unit-100)*5
    print("amount:",amount)
else:
    amount=500+(unit-200)*10
    print("amount:",amount)
```

### Accept the following from the user and calculate the percentage of class attended.

    a=number of working days
    b=total number of days for absent
  After calculating percentage show that if the percentage is >70 then students are eligible for bonus otherwise not.
    

```python
a=int(input("Enter number of working days: "))
b=int(input("Enter total number of days for absent: "))
c=a-b

percentage=(c/a)*100
print("percentage is:",percentage,"%")

if percentage>70:
    print("eligible for bonus")
else:
    print("Not eligible")
```

### Write a python program to take the cost price of bike from user and disply the road tax to be paid based on purchase price of bike.

```python
price = int(input("Enter cost price:"))

if price>=200000:
    tax=15*price/100
    print("The tax on bike is: ",tax)
elif price>=100000 and price<=200000:
    tax=10*price/100
    print("The tax on bike is: ",tax)
elif price<=100000:
    tax=5*price/100
    print("The tax on bike is: ",tax)
else:
    print("ThnkYou")
```
