## Different categories for User Defined Functions. 

### 1.Function without Parameter and no return type: 

```python
def wish():
    name=input("Enter your name:")
    print(name)
wish()
```

### 2.Function with Parameter and no return type:

```python
name=input("Enter your name:")

def wish(name):
    print(name)
wish(name)
```

#### Default Argument: 

```python
def wish(name):
    print(name)
wish("qwer")
```

### 3.Function without Parameter and with return type: 

```python
def wish():
    name=input("Enter your name:")
    return name
wish()
```

###  4.Function with Parameter and with return type:

```python
def wish(name):
    return name

name=input("Enter your name:")
wish(name)
```

## Types of Arguments: 

### 1.Possitional Argument

```python
def sub(a,b):
    return a-b
print(sub(10,20))
print(sub(20,10))
```

### 2.Keyword Argument 

```python
def wish(name,msg):
    print("Hello",name,msg)

wish(name="Dev",msg="Good Morning")
wish("Dev","Good Morning")
wish(msg="Good Morning",name="Dev")
wish("Good Morning","Dev")
wish("Dev",msg="Good Morning")
#wish(name="Dev","Good Morning") Error:Possitional 
```
